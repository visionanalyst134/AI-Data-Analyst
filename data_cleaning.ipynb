# 1.Import Libraries

```python
import pandas as pd
```

# 2. Load Dataset

```python
df = pd.read_csv("raw_csv_data/data.csv", encoding="cp1252")
```

# 4. Remove Duplicates

```python
# duplicate rows remove
df = df.drop_duplicates()
```

```python
df.duplicated().sum()
```

# 5. Fix Date Data Types

```python
# InvoiceDate ko datetime me convert
df["InvoiceDate"] = pd.to_datetime(df["InvoiceDate"])
```

```python
df.info()
```

# 6. Handle Missing Values

```python
# Description null rows remove
df = df.dropna(subset=["Description"])
```

```python
df.shape # duplicates + description rows are delete...
```

```python
df.isnull().sum()
```

# 7.Investigate Invalid Quantity / UnitPrice

# Deal with negative and invalid Unit Price

```python
df[df["UnitPrice"] <= 0].shape
```

```python
# two types of unit price 
#1.-ve unit price
#2.0.0 unit price
```

```python
#1.-ve unit price
df[df["UnitPrice"] < 0]
```

```python
#2. 0.0 unit price
df[df["UnitPrice"] <= 0]
```

# Deal with negative Quantity and invoice

```python
#total invoice with -ve
df[df["Quantity"] <= 0].shape
```

```python
# Negative Quantity investigate
df[df["InvoiceNo"].str.startswith("C")].shape
```

```python
df[df["InvoiceNo"].str.startswith("C")].head(5)
```

```python
# Negative Quantity but non-C invoices 
df[
    (df["Quantity"] <= 0) &
    (~df["InvoiceNo"].str.startswith("C"))
].shape
```

# Note:-
# Rows with cancelled invoices (InvoiceNo starting with "C") and negative quantities were removed
# because they do not represent valid sales transactions.

# For UnitPrice, negative values were mainly linked to invalid or adjustment-type records,
# so those rows were also removed.

# 8.df_clean table----> by Remove Invalid and -ve Sales Rows..

```python
df_clean = df[
    (df["Quantity"] > 0) &
    (df["UnitPrice"] > 0)
].copy()
```

```python
df_clean.shape
```

```python
df_clean.head(10)
```

```python
df_clean.size
```

```python
df_clean.isnull().sum()
```

# df_customer table ---> for customer level analysis..

```python
df_customer = df_clean.dropna(subset=["CustomerID"]).copy()
```

```python
df_customer.head(10)
```

```python
df_clean.isnull().sum()
```

```python
df_customer.shape
```


# 9.Create Revenue and Date Features

```python
df_clean["Revenue"] = df_clean["Quantity"] * df_clean["UnitPrice"]
```

```python
df_clean.head(5)
```

# create date fields/columns using invoice date..

```python
df_clean["Year"] = df_clean["InvoiceDate"].dt.year
df_clean["Month"] = df_clean["InvoiceDate"].dt.month
df_clean["MonthName"] = df_clean["InvoiceDate"].dt.month_name()
df_clean["Day"] = df_clean["InvoiceDate"].dt.day
```

```python
df_clean.head()
```

```python
# Data type change
df_customer["CustomerID"] = df_customer["CustomerID"].astype("int64")
```
